```python
import numpy as np
import matplotlib.pyplot as plt

a = 1
b = 1
c = 1
d = 1

x0 = 2
y0 = 1

t0 = 0
t_end = 100
h = 0.01

def f(t, X):
    x, y = X
    dxdt = a * x - b * x * y
    dydt = c * x * y - d * y
    return np.array([dxdt, dydt])

def runge_kutta_4th_order(f, t0, X0, t_end, h):
    t_values = [t0]
    X_values = [X0]
    
    t = t0
    X = X0
    
    while t < t_end:
        k1 = f(t, X)
        k2 = f(t + h / 2, X + h / 2 * k1)
        k3 = f(t + h / 2, X + h / 2 * k2)
        k4 = f(t + h, X + h * k3)
        
        X = X + h / 6 * (k1 + 2 * k2 + 2 * k3 + k4)
        t = t + h
        
        t_values.append(t)
        X_values.append(X)
    
    return np.array(t_values), np.array(X_values)

t_values, X_values = runge_kutta_4th_order(f, t0, np.array([x0, y0]), t_end, h)

x_values = X_values[:, 0]
y_values = X_values[:, 1]

plt.figure(figsize=(10, 5))


plt.subplot(1, 2, 2)
plt.plot(x_values, y_values)
plt.xlabel('Beute (x)')
plt.ylabel('Räuber (y)')
plt.title('Phasenraumdiagramm')

plt.tight_layout()
plt.show()
```
